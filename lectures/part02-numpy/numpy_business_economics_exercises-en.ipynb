## Worked Example 1.3 — Moving Average

For a window of width \(w\), the simple moving average is:

$$
SMA_t
=
\frac{1}{w}
\sum_{i=0}^{w-1} x_{t-i}
$$

````python
example_series = np.array([10, 12, 11, 15, 18], dtype=float)

window_size = 3
weights = np.ones(window_size) / window_size

example_sma = np.convolve(
    example_series,
    weights,
    mode="valid"
)

print("3-period SMA:", example_sma)
````

## Worked Example 2.2 — Arc Price Elasticity

For adjacent observations:

$$
E_d
=
\frac{
\frac{Q_2-Q_1}{(Q_2+Q_1)/2}
}{
\frac{P_2-P_1}{(P_2+P_1)/2}
}
$$

````python
p = np.array([10.0, 11.0, 12.0])
q = np.array([100.0, 92.0, 80.0])

q_mid_change = (
    np.diff(q)
    / ((q[1:] + q[:-1]) / 2)
)

p_mid_change = (
    np.diff(p)
    / ((p[1:] + p[:-1]) / 2)
)

example_elasticity = q_mid_change / p_mid_change

print("Elasticity:", example_elasticity)
````

# Section 5 — Forecast Evaluation

## Worked Example 5.1 — Forecast Error

$$
MAE
=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

$$
RMSE
=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

````python
actual = np.array([100, 120, 110], dtype=float)
forecast = np.array([98, 125, 108], dtype=float)

error = actual - forecast
mae = np.mean(np.abs(error))
rmse = np.sqrt(np.mean(error ** 2))

print("MAE:", mae)
print("RMSE:", rmse)
````
