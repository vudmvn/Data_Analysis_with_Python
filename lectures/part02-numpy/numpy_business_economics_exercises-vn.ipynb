=
\frac{100}{n}
\sum_{i=1}^{n}
\left|
\frac{y_i-\hat{y}_i}{y_i}
\right|
$$

## Ví dụ mẫu 1.3 — Trung bình trượt

Với cửa sổ có độ rộng \(w\), trung bình trượt đơn giản được tính bằng:

$$
SMA_t
=
\frac{1}{w}
\sum_{i=0}^{w-1} x_{t-i}
$$

````python
example_series = np.array([10, 12, 11, 15, 18], dtype=float)

window_size = 3
weights = np.ones(window_size) / window_size

example_sma = np.convolve(
    example_series,
    weights,
    mode="valid"
)

print("SMA ba kỳ:", example_sma)
````

## Ví dụ mẫu 2.2 — Độ co giãn vòng cung của cầu theo giá

Với hai quan sát liền kề:

$$
E_d
=
\frac{
\frac{Q_2-Q_1}{(Q_2+Q_1)/2}
}{
\frac{P_2-P_1}{(P_2+P_1)/2}
}
$$

````python
p = np.array([10.0, 11.0, 12.0])
q = np.array([100.0, 92.0, 80.0])

q_mid_change = (
    np.diff(q)
    / ((q[1:] + q[:-1]) / 2)
)

p_mid_change = (
    np.diff(p)
    / ((p[1:] + p[:-1]) / 2)
)

example_elasticity = q_mid_change / p_mid_change

print("Độ co giãn:", example_elasticity)
````

# Phần 3 — Phân tích chi phí, lợi nhuận và điểm hòa vốn

## Ví dụ mẫu 3.1 — Lợi nhuận

$$
\text{Lợi nhuận}
=
\text{Doanh thu}
-
\text{Tổng chi phí}
$$

$$
\text{Tổng chi phí}
=
\text{Chi phí cố định}
+
\text{Chi phí biến đổi trên đơn vị}
\times
Q
$$

````python
example_price = np.array([10, 12, 14], dtype=float)
example_demand = np.array([1000, 850, 700], dtype=float)

fixed_cost = 3000.0
variable_cost = 6.0

example_revenue = example_price * example_demand
example_total_cost = fixed_cost + variable_cost * example_demand
example_profit = example_revenue - example_total_cost

print("Lợi nhuận:", example_profit)
````

# Phần 5 — Đánh giá dự báo

## Ví dụ mẫu 5.1 — Sai số dự báo

$$
MAE
=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

$$
RMSE
=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}

}
$$

````python
actual = np.array([100, 120, 110], dtype=float)
forecast = np.array([98, 125, 108], dtype=float)

error = actual - forecast
mae = np.mean(np.abs(error))
rmse = np.sqrt(np.mean(error ** 2))

print("MAE:", mae)
print("RMSE:", rmse)
````
