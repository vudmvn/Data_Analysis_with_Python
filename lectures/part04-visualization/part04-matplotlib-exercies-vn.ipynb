### Sản phẩm cần nộp

Một **Business Performance Dashboard** bằng Matplotlib và 3–5 nhận xét quản trị.

### 1.1 Tạo dữ liệu

````python
months = pd.date_range("2025-01-01", periods=12, freq="MS")
regions = ["North", "Central", "South", "Online"]
categories = ["Electronics", "Home", "Fashion"]

rows = []

for month_idx, month in enumerate(months):
    seasonal = 1 + 0.12*np.sin((month_idx/12)*2*np.pi)
    trend = 1 + month_idx*0.025

    for region in regions:
        for category in categories:
            region_factor = {
                "North": 1.00,
                "Central": 0.82,
                "South": 1.12,
                "Online": 1.28
            }[region]

            category_factor = {
                "Electronics": 1.35,
                "Home": 1.00,
                "Fashion": 0.88
            }[category]

            revenue = (
                420
                * seasonal
                * trend
                * region_factor
                * category_factor
                * np.random.normal(1, 0.07)
            )

            margin = {
                "Electronics": 0.13,
                "Home": 0.20,
                "Fashion": 0.27
            }[category] + np.random.normal(0, 0.015)

            profit = revenue * margin
            orders = int(revenue * np.random.uniform(1.8, 2.5))

            rows.append([
                month, region, category,
                revenue, profit, orders
            ])

sales = pd.DataFrame(
    rows,
    columns=["Month", "Region", "Category", "Revenue", "Profit", "Orders"]
)

# Một cú sốc mô phỏng ở tháng 8 đối với khu vực Central
mask = (sales["Month"] == "2025-08-01") & (sales["Region"] == "Central")
sales.loc[mask, "Revenue"] *= 0.72
sales.loc[mask, "Profit"] *= 0.70

sales.head()
````

> Dữ liệu dưới đây là **synthetic data**, không phải số liệu thực tế của một quốc gia.

### 2.1 Tạo dữ liệu kinh tế

````python
quarters = pd.period_range("2021Q1", "2025Q4", freq="Q")
t = np.arange(len(quarters))

inflation = (
    2.1
    + 4.8*np.exp(-((t-7)/3.1)**2)
    + 0.25*np.random.randn(len(t))
)

unemployment = (
    5.4
    - 0.16*t
    + 1.7*np.exp(-((t-13)/2.4)**2)
    + 0.18*np.random.randn(len(t))
)

gdp_growth = (
    4.8
    + 0.7*np.sin(t/2)
    - 3.2*np.exp(-((t-13)/2.0)**2)
    + 0.35*np.random.randn(len(t))
)

economy = pd.DataFrame({
    "Quarter": quarters.astype(str),
    "Inflation": inflation,
    "Unemployment": unemployment,
    "GDP_Growth": gdp_growth
})

economy.round(2)
````

# MINI-PROJECT 3 — DATA SCIENCE
## Exploratory Data Analysis cho Customer Churn

### Bối cảnh

Một công ty subscription muốn tìm hiểu khách hàng rời bỏ dịch vụ (`Churn = 1`).

Trước khi xây dựng machine learning model, Data Scientist cần thực hiện **EDA** để trả lời:

- Churn có mất cân bằng không?
- Khách hàng churn có tenure khác khách hàng không churn không?
- Monthly Charge có liên hệ với churn không?
- Customer profile nào đáng chú ý?

### Mục tiêu

Sử dụng Matplotlib để tạo một **EDA report bằng hình ảnh** trước khi modeling.

### 3.1 Tạo dữ liệu churn

````python
n = 1200

tenure = np.random.gamma(shape=2.2, scale=11, size=n)
tenure = np.clip(tenure, 1, 72)

monthly_charge = np.random.normal(65, 18, n)
monthly_charge = np.clip(monthly_charge, 20, 120)

support_calls = np.random.poisson(1.8, n)

contract = np.random.choice(
    ["Monthly", "Annual", "Two-year"],
    size=n,
    p=[0.55, 0.30, 0.15]
)

logit = (
    -1.7
    + 0.020*(monthly_charge - 60)
    - 0.035*(tenure - 20)
    + 0.32*support_calls
    + np.where(contract == "Monthly", 0.65, 0)
    + np.where(contract == "Two-year", -0.75, 0)
)

p_churn = 1 / (1 + np.exp(-logit))
churn = np.random.binomial(1, p_churn)

customers = pd.DataFrame({
    "Tenure": tenure,
    "MonthlyCharge": monthly_charge,
    "SupportCalls": support_calls,
    "Contract": contract,
    "Churn": churn
})

customers.head()
````

### 4.1 Tạo dữ liệu pricing

````python
prices = np.arange(20, 101, 5)

segments = {
    "Budget":  {"base": 1600, "sensitivity": 13.5},
    "Mainstream": {"base": 1500, "sensitivity": 9.0},
    "Premium": {"base": 1100, "sensitivity": 4.8}
}

pricing_rows = []

for segment, pars in segments.items():
    for price in prices:
        demand = (
            pars["base"]
            - pars["sensitivity"] * price
            + np.random.normal(0, 35)
        )
        demand = max(demand, 20)
        revenue = price * demand

        pricing_rows.append([segment, price, demand, revenue])

pricing = pd.DataFrame(
    pricing_rows,
    columns=["Segment", "Price", "Demand", "Revenue"]
)

pricing.head()
````
