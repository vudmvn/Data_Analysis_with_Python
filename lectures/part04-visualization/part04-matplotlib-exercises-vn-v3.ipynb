### 1.1 Tạo dữ liệu kinh doanh mô phỏng

Trong mini-project này, ta chưa sử dụng dữ liệu thật từ doanh nghiệp.
Thay vào đó, notebook tạo một **dataset mô phỏng** có cấu trúc gần giống dữ liệu bán hàng thực tế.

Mỗi dòng dữ liệu tương ứng với một tổ hợp:

> **Tháng × Khu vực × Nhóm sản phẩm**

Ví dụ:

| Month | Region | Category | Revenue | Profit | Orders |
|---|---|---|---:|---:|---:|
| 2025-01 | North | Electronics | ... | ... | ... |
| 2025-01 | North | Home | ... | ... | ... |
| 2025-01 | Central | Electronics | ... | ... | ... |

Với:

- 12 tháng;
- 4 khu vực;
- 3 nhóm sản phẩm;

ta có:

\[
12 \times 4 \times 3 = 144
\]

quan sát.

#### Ý tưởng mô phỏng

Ta muốn dữ liệu **không hoàn toàn ngẫu nhiên**, mà có một số đặc điểm kinh doanh để
sau này biểu đồ có ý nghĩa:

1. **Trend:** doanh thu tăng nhẹ theo thời gian.
2. **Seasonality:** doanh thu dao động theo mùa.
3. **Region effect:** mỗi khu vực có quy mô doanh thu khác nhau.
4. **Category effect:** mỗi nhóm sản phẩm có mức doanh thu khác nhau.
5. **Random noise:** dữ liệu thực tế luôn có biến động nhỏ ngoài mô hình.
6. **Business shock:** khu vực `Central` bị giảm doanh thu trong tháng 8.

Nhờ vậy, khi trực quan hóa, ta có thể quan sát được **trend, khác biệt giữa nhóm và anomaly**.

#### Lưu ý cho người học

Bạn **không cần học thuộc công thức tạo Revenue** trong phần này.

Mục đích của đoạn code chỉ là tạo ra một dataset có:

- cấu trúc rõ ràng;
- xu hướng;
- khác biệt giữa các nhóm;
- một vài biến động để thực hành trực quan hóa.

Trọng tâm của project bắt đầu từ bước **đọc, tổng hợp và trực quan hóa `sales`**.


````python

# ------------------------------------------------------------
# BƯỚC 1: Xác định các chiều của dataset
# ------------------------------------------------------------

# Tạo 12 tháng, bắt đầu từ tháng 01/2025.
# freq="MS" nghĩa là lấy ngày đầu tiên (Month Start) của mỗi tháng.
months = pd.date_range("2025-01-01", periods=12, freq="MS")

# Các khu vực/kênh bán hàng.
regions = ["North", "Central", "South", "Online"]

# Các nhóm sản phẩm.
categories = ["Electronics", "Home", "Fashion"]


# ------------------------------------------------------------
# BƯỚC 2: Tạo từng dòng dữ liệu
# ------------------------------------------------------------

# rows sẽ chứa các dòng trước khi chuyển thành DataFrame.
rows = []

# enumerate() trả về cả vị trí month_idx và giá trị month.
for month_idx, month in enumerate(months):

    # seasonal tạo dao động theo mùa.
    # Giá trị nằm xung quanh 1.
    seasonal = 1 + 0.12 * np.sin((month_idx / 12) * 2 * np.pi)

    # trend làm doanh thu nền tăng khoảng 2.5% qua mỗi bước thời gian.
    trend = 1 + month_idx * 0.025

    # Với mỗi tháng, tạo dữ liệu cho từng khu vực...
    for region in regions:

        # ...và từng nhóm sản phẩm.
        for category in categories:

            # Hệ số khu vực:
            # Online có quy mô doanh thu cao hơn,
            # Central thấp hơn mức cơ sở North.
            region_factor = {
                "North": 1.00,
                "Central": 0.82,
                "South": 1.12,
                "Online": 1.28
            }[region]

            # Hệ số nhóm sản phẩm:
            # Electronics có doanh thu nền cao hơn,
            # Fashion thấp hơn mức cơ sở Home.
            category_factor = {
                "Electronics": 1.35,
                "Home": 1.00,
                "Fashion": 0.88
            }[category]

            # Revenue được hình thành từ:
            # doanh thu cơ sở × mùa vụ × xu hướng
            # × khu vực × nhóm sản phẩm × nhiễu ngẫu nhiên.
            revenue = (
                420
                * seasonal
                * trend
                * region_factor
                * category_factor
                * np.random.normal(1, 0.07)
            )

            # Mỗi category có biên lợi nhuận cơ sở khác nhau.
            # Ta thêm một lượng nhiễu nhỏ để dữ liệu tự nhiên hơn.
            margin = {
                "Electronics": 0.13,
                "Home": 0.20,
                "Fashion": 0.27
            }[category] + np.random.normal(0, 0.015)

            # Profit = Revenue × Margin.
            profit = revenue * margin

            # Số đơn hàng được tạo có quan hệ dương với Revenue.
            orders = int(revenue * np.random.uniform(1.8, 2.5))

            # Thêm một dòng vào list rows.
            rows.append([
                month,
                region,
                category,
                revenue,
                profit,
                orders
            ])


# ------------------------------------------------------------
# BƯỚC 3: Chuyển list thành DataFrame
# ------------------------------------------------------------

sales = pd.DataFrame(
    rows,
    columns=[
        "Month",
        "Region",
        "Category",
        "Revenue",
        "Profit",
        "Orders"
    ]
)


# ------------------------------------------------------------
# BƯỚC 4: Tạo một "business shock" để có anomaly khi phân tích
# ------------------------------------------------------------

# Chọn các dòng thuộc tháng 08/2025 và khu vực Central.
mask = (
    (sales["Month"] == "2025-08-01")
    & (sales["Region"] == "Central")
)

# Giả lập một sự cố khiến Revenue và Profit giảm mạnh.
sales.loc[mask, "Revenue"] *= 0.72
sales.loc[mask, "Profit"] *= 0.70


# Xem 5 dòng đầu tiên.
sales.head()
````

### 2.1 Tạo dữ liệu kinh tế mô phỏng

#### Vì sao tạo dữ liệu theo cách này?

Ta muốn các chuỗi thời gian có một số hiện tượng để thực hành đọc biểu đồ:

- Inflation có một giai đoạn tăng mạnh rồi giảm.

- Mỗi chuỗi có thêm một lượng **random noise** nhỏ.

Như vậy, dữ liệu có tính chất giống một time series kinh tế đơn giản hơn là một dãy số ngẫu nhiên hoàn toàn.

---

#### Các câu lệnh quan trọng

**`pd.period_range()`**

```python
quarters = pd.period_range("2021Q1", "2025Q4", freq="Q")
```

Tạo các quý liên tục từ `2021Q1` đến `2025Q4`.

**`np.exp()`**

Trong dataset này, `np.exp()` được dùng để tạo một **đỉnh hoặc cú sốc tạm thời**.

Ví dụ về ý tưởng:

- trước cú sốc: tác động nhỏ;
- gần trung tâm cú sốc: tác động lớn;

Bạn **không cần học thuộc công thức toán học này** để làm bài Matplotlib.

---

**`np.random.randn()`**

Thêm nhiễu ngẫu nhiên xung quanh xu hướng chính.

Dữ liệu kinh tế thực tế hiếm khi di chuyển theo một đường hoàn toàn trơn.

---

**`pd.DataFrame({...})`**

Cuối cùng, các chuỗi được ghép thành một DataFrame:

```python
economy = pd.DataFrame({
    "Quarter": ...,
    "Inflation": ...,
    "Unemployment": ...,
    "GDP_Growth": ...
})
```

Từ bước này trở đi, ta chỉ cần sử dụng DataFrame `economy` để phân tích.


````python

# ------------------------------------------------------------
# BƯỚC 1: Tạo trục thời gian theo quý
# ------------------------------------------------------------

# Tạo các quý liên tục từ 2021Q1 đến 2025Q4.
quarters = pd.period_range("2021Q1", "2025Q4", freq="Q")

# t = 0, 1, 2, ..., dùng làm chỉ số thời gian trong các công thức mô phỏng.
t = np.arange(len(quarters))


# ------------------------------------------------------------
# BƯỚC 2: Mô phỏng Inflation
# ------------------------------------------------------------

# Inflation gồm:
# - mức cơ sở khoảng 2.1;
# - một đợt tăng mạnh tập trung quanh t = 7;
# - nhiễu nhỏ để dữ liệu không quá "đẹp".
inflation = (
    2.1
    + 4.8 * np.exp(-((t - 7) / 3.1) ** 2)
    + 0.25 * np.random.randn(len(t))
)


# ------------------------------------------------------------
# BƯỚC 3: Mô phỏng Unemployment
# ------------------------------------------------------------

# Unemployment gồm:
# - mức ban đầu khoảng 5.4;
# - xu hướng giảm dần theo thời gian;
# - một giai đoạn tăng tạm thời quanh t = 13;
# - nhiễu nhỏ.
unemployment = (
    5.4
    - 0.16 * t
    + 1.7 * np.exp(-((t - 13) / 2.4) ** 2)
    + 0.18 * np.random.randn(len(t))
)


# ------------------------------------------------------------
# BƯỚC 4: Mô phỏng GDP Growth
# ------------------------------------------------------------

# GDP Growth gồm:
# - mức cơ sở khoảng 4.8;
# - dao động nhẹ theo chu kỳ;
# - một cú giảm rõ rệt quanh t = 13;
# - nhiễu ngẫu nhiên nhỏ.
gdp_growth = (
    4.8
    + 0.7 * np.sin(t / 2)
    - 3.2 * np.exp(-((t - 13) / 2.0) ** 2)
    + 0.35 * np.random.randn(len(t))
)


# ------------------------------------------------------------
# BƯỚC 5: Ghép các biến thành DataFrame
# ------------------------------------------------------------

economy = pd.DataFrame({
    # Chuyển Period thành chuỗi như "2021Q1", "2021Q2", ...
    "Quarter": quarters.astype(str),

    # Ba chỉ số kinh tế vừa tạo.
    "Inflation": inflation,
    "Unemployment": unemployment,
    "GDP_Growth": gdp_growth
})


# Làm tròn khi hiển thị để bảng dễ đọc hơn.
# Dữ liệu gốc trong DataFrame không bị thay đổi bởi lệnh này.
economy.round(2)
````

- `SupportCalls`: gọi hỗ trợ nhiều → xác suất churn có xu hướng tăng;
- `Contract`: hợp đồng tháng có rủi ro cao hơn; hợp đồng hai năm thấp hơn.

Đây chỉ là **quy tắc mô phỏng**, không phải kết luận về churn trong doanh nghiệp thực tế.

---

#### Các hàm tạo dữ liệu được sử dụng

**`np.random.gamma()`**

Dùng để tạo `Tenure`.

Phân phối gamma giúp tạo nhiều khách hàng có tenure thấp/trung bình và ít khách hàng có tenure rất cao.

---

**`np.random.normal()`**

Dùng để tạo `MonthlyCharge` tập trung xung quanh một mức trung bình.

---

**`np.random.poisson()`**

Dùng để tạo số lần `SupportCalls`.

Poisson thường phù hợp để mô phỏng các biến dạng **số lần một sự kiện xảy ra**.

---

**`np.random.choice()`**

Dùng để chọn loại hợp đồng:

```python
["Monthly", "Annual", "Two-year"]
```

với các xác suất khác nhau.

---

#### Từ đặc trưng đến xác suất Churn

Đoạn code tạo một điểm số `logit`.

Bạn có thể hiểu đơn giản:

> `logit` càng lớn → khách hàng càng có xu hướng churn.

Sau đó:

```python
p_churn = 1 / (1 + np.exp(-logit))
```

chuyển điểm số đó thành xác suất từ `0` đến `1`.

Cuối cùng:

```python
churn = np.random.binomial(1, p_churn)
```

tạo giá trị `0/1` dựa trên xác suất vừa tính.

> Phần logistic này chỉ nhằm tạo dữ liệu có pattern để EDA. Không yêu cầu người học phải hiểu Logistic Regression ở bài Matplotlib này.


````python

# ------------------------------------------------------------
# BƯỚC 1: Xác định số khách hàng
# ------------------------------------------------------------

n = 1200


# ------------------------------------------------------------
# BƯỚC 2: Tạo các đặc trưng khách hàng
# ------------------------------------------------------------

# Tenure = số tháng sử dụng dịch vụ.
# Gamma tạo nhiều quan sát ở vùng thấp/trung bình và ít quan sát rất lớn.
tenure = np.random.gamma(
    shape=2.2,
    scale=11,
    size=n
)

# Giới hạn tenure trong khoảng hợp lý từ 1 đến 72 tháng.
tenure = np.clip(tenure, 1, 72)


# MonthlyCharge tập trung quanh 65.
monthly_charge = np.random.normal(
    loc=65,
    scale=18,
    size=n
)

# Giới hạn mức phí trong khoảng 20 đến 120.
monthly_charge = np.clip(monthly_charge, 20, 120)


# SupportCalls là số lần gọi hỗ trợ.
support_calls = np.random.poisson(
    lam=1.8,
    size=n
)


# Loại hợp đồng.
# p quy định xác suất xuất hiện của từng loại:
# Monthly 55%, Annual 30%, Two-year 15%.
contract = np.random.choice(
    ["Monthly", "Annual", "Two-year"],
    size=n,
    p=[0.55, 0.30, 0.15]
)


# ------------------------------------------------------------
# BƯỚC 3: Tạo xác suất Churn có liên quan đến các đặc trưng
# ------------------------------------------------------------

# Có thể hiểu logit là một "điểm rủi ro churn".
#
# - MonthlyCharge cao hơn  -> điểm tăng.
# - Tenure cao hơn         -> điểm giảm.
# - SupportCalls nhiều hơn -> điểm tăng.
# - Monthly contract       -> điểm tăng.
# - Two-year contract      -> điểm giảm.
logit = (
    -1.7
    + 0.020 * (monthly_charge - 60)
    - 0.035 * (tenure - 20)
    + 0.32 * support_calls
    + np.where(contract == "Monthly", 0.65, 0)
    + np.where(contract == "Two-year", -0.75, 0)
)


# Chuyển điểm logit thành xác suất từ 0 đến 1.
p_churn = 1 / (1 + np.exp(-logit))


# Dựa trên xác suất của từng khách hàng,
# sinh Churn = 0 hoặc Churn = 1.
churn = np.random.binomial(
    n=1,
    p=p_churn
)


# ------------------------------------------------------------
# BƯỚC 4: Ghép các biến thành DataFrame
# ------------------------------------------------------------

customers = pd.DataFrame({
    "Tenure": tenure,
    "MonthlyCharge": monthly_charge,
    "SupportCalls": support_calls,
    "Contract": contract,
    "Churn": churn
})


# Kiểm tra 5 khách hàng đầu tiên.
customers.head()
````

### 4.1 Tạo dữ liệu Pricing mô phỏng

Trong mini-project này, ta mô phỏng phản ứng của ba phân khúc khách hàng
khi doanh nghiệp thử các mức giá khác nhau.

Ba segment:

- `Budget`
- `Mainstream`
- `Premium`

Giá được thử từ **20 đến 100**, mỗi lần tăng **5 đơn vị**.

---

#### Ý tưởng kinh tế cơ bản

Ta mô phỏng quan hệ:

\[
Demand = BaseDemand - Sensitivity \times Price + Noise
\]

Có thể hiểu:

- `BaseDemand`: quy mô nhu cầu cơ sở của segment;
- `Sensitivity`: mức độ demand giảm khi price tăng;
- `Noise`: biến động ngẫu nhiên do các yếu tố khác.

Ví dụ:

```python
"Budget": {"base": 1600, "sensitivity": 13.5}
```

nghĩa là:

- Budget có thị trường cơ sở khá lớn;
- nhưng demand giảm nhanh khi price tăng.

Trong khi đó:

```python
"Premium": {"base": 1100, "sensitivity": 4.8}
```

có:

- base demand thấp hơn;
- nhưng ít nhạy cảm với giá hơn.

Đây là dữ liệu mô phỏng để thể hiện **price sensitivity**, không phải hàm cầu được ước lượng từ dữ liệu thực tế.

---

#### Các cấu trúc code quan trọng

**`np.arange(20, 101, 5)`**

Tạo các mức giá:

```text
20, 25, 30, ..., 100
```

Lưu ý: cận trên của `np.arange()` không được lấy, vì vậy dùng `101` để có giá `100`.

---

**Dictionary lồng nhau**

```python
segments = {
    "Budget": {"base": ..., "sensitivity": ...},
    ...
}
```

Dictionary giúp lưu các tham số riêng của từng segment.

Ta có thể truy cập:

```python
segments["Budget"]["sensitivity"]
```

---

**Vòng lặp**

```python
for segment, pars in segments.items():
    for price in prices:
        ...
```

Code tạo một quan sát cho **mỗi Segment × Price**.

---

#### Tạo Demand

```python
demand = (
    pars["base"]
    - pars["sensitivity"] * price
    + np.random.normal(0, 35)
)
```

Demand giảm khi Price tăng, nhưng có thêm một lượng biến động ngẫu nhiên.

Sau đó:

```python
demand = max(demand, 20)
```

đảm bảo demand không xuống dưới 20.

---

#### Tính Revenue

Sau khi có Price và Demand:

\[
Revenue = Price \times Demand
\]

Trong code:

```python
revenue = price * demand
```

Điểm quan trọng của mini-project là:

> **Demand có thể giảm khi Price tăng nhưng Revenue chưa chắc giảm ngay.**

Đó là lý do ta sẽ trực quan hóa cả `Price → Demand` và `Price → Revenue`.


````python

# ------------------------------------------------------------
# BƯỚC 1: Tạo các mức giá cần thử
# ------------------------------------------------------------

# Các mức giá: 20, 25, 30, ..., 100.
prices = np.arange(20, 101, 5)


# ------------------------------------------------------------
# BƯỚC 2: Định nghĩa đặc điểm của từng phân khúc
# ------------------------------------------------------------

# base:
#   mức nhu cầu cơ sở của segment.
#
# sensitivity:
#   mức demand giảm khi price tăng thêm 1 đơn vị.
#
# sensitivity càng lớn -> segment càng nhạy cảm với giá.
segments = {
    "Budget": {
        "base": 1600,
        "sensitivity": 13.5
    },
    "Mainstream": {
        "base": 1500,
        "sensitivity": 9.0
    },
    "Premium": {
        "base": 1100,
        "sensitivity": 4.8
    }
}


# ------------------------------------------------------------
# BƯỚC 3: Tạo dữ liệu cho từng Segment × Price
# ------------------------------------------------------------

pricing_rows = []

# .items() trả về:
# - segment: tên phân khúc;
# - pars: dictionary chứa base và sensitivity.
for segment, pars in segments.items():

    # Thử từng mức giá.
    for price in prices:

        # Demand giảm tuyến tính khi Price tăng.
        # np.random.normal(0, 35) thêm biến động ngẫu nhiên.
        demand = (
            pars["base"]
            - pars["sensitivity"] * price
            + np.random.normal(0, 35)
        )

        # Không cho demand nhỏ hơn 20.
        demand = max(demand, 20)

        # Revenue = Price × Demand.
        revenue = price * demand

        # Lưu kết quả của một Segment × Price.
        pricing_rows.append([
            segment,
            price,
            demand,
            revenue
        ])


# ------------------------------------------------------------
# BƯỚC 4: Chuyển kết quả thành DataFrame
# ------------------------------------------------------------

pricing = pd.DataFrame(
    pricing_rows,
    columns=[
        "Segment",
        "Price",
        "Demand",
        "Revenue"
    ]
)


# Xem 5 dòng đầu tiên.
pricing.head()
````

### Dẫn dắt

Demand giảm khi tăng giá không đồng nghĩa Revenue chắc chắn giảm.
